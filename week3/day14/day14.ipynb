```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
```

## Task 2

```python
sns.histplot(data=tips, x="total_bill", hue="time")
plt.title("توزيع قيمة الفاتورة حسب الوجبة")
plt.show()
```

```python
# 1. أتوقع أن العشاء فواتيره أكبر
sns.histplot(data=tips, x="total_bill", hue="time")
plt.title("توزيع قيمة الفاتورة حسب الوجبة"); plt.show()

# 2. أتوقع أن السبت والأحد أعلى بقشيش
sns.boxplot(data=tips, x="day", y="tip")
plt.title("البقشيش حسب اليوم"); plt.show()

# 3. أتوقع أن السبت أكثر أيام الفواتير
sns.countplot(data=tips, x="day", hue="smoker")
plt.title("عدد الفواتير في كل يوم حسب التدخين"); plt.show()

# 4. أتوقع ارتباطًا موجبًا
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time", size="size")
plt.title("الفاتورة مقابل البقشيش"); plt.show()
print("الارتباط:", tips["total_bill"].corr(tips["tip"]).round(2))

# 5. أتوقع أن الفرق بين الأيام صغير
sns.barplot(data=tips, x="day", y="tip")
plt.title("متوسط البقشيش لكل يوم"); plt.show()

# 6. أتوقع أن الفاتورة والبقشيش أعلى ارتباط
sns.heatmap(tips[["total_bill","tip","size"]].corr(), annot=True)
plt.title("مصفوفة الارتباط"); plt.show()
```

```python
# Task 2 conclusions
conclusions = """
1. Dinner bills are larger and more frequent than lunch.
2. Sunday has the highest median tip.
3. Saturday has the most bills.
4. Yes, higher bill means higher tip. Correlation = 0.68.
5. The difference between days is probably not real because confidence intervals overlap.
6. Highest correlation: total_bill and tip (0.68), then total_bill and size (0.6), lowest: tip and size (0.49).
"""
print(conclusions)
```

```python
import numpy as np
import pandas as pd
rng = np.random.default_rng(7)
n = 150
species = rng.choice(["Fire", "Ice", "Storm"], n, p=[0.4, 0.35, 0.25])
base_wing = {"Fire": 14, "Ice": 11, "Storm": 18}
base_heat = {"Fire": 900, "Ice": 80, "Storm": 300}
dragons = pd.DataFrame({
    "dragon_id": [f"D{i:03d}" for i in range(n)],
    "species": species,
    "wingspan_m": [round(base_wing[s] + rng.normal(0, 2), 1) for s in species],
    "fire_temp_c": [round(base_heat[s] + rng.normal(0, 60)) for s in species],
    "age_years": rng.integers(20, 400, n),
})
dragons["weight_t"] = (dragons["wingspan_m"] * 0.6 + rng.normal(0, 0.8, n)).round(1)
dragons.loc[37, "wingspan_m"] = 140.0
dragons.head()
```

```python
print(dragons["wingspan_m"].describe())
print(dragons[dragons["wingspan_m"] > 30])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. countplot: how many dragons of each species
sns.countplot(data=dragons, x="species")
plt.title("Dragons per species"); plt.show()

# 2. boxplot BEFORE fixing: the weird dragon appears as an outlier
sns.boxplot(data=dragons, x="species", y="wingspan_m")
plt.title("Wingspan by species (before fix)"); plt.show()

# fix the error: D037 has wingspan 140 -> 14.0
dragons.loc[37, "wingspan_m"] = 14.0

sns.boxplot(data=dragons, x="species", y="wingspan_m")
plt.title("Wingspan by species (after fix)"); plt.show()

# 3. scatterplot: wingspan vs fire temperature, colored by species
sns.scatterplot(data=dragons, x="wingspan_m", y="fire_temp_c", hue="species")
plt.title("Wingspan vs fire temperature"); plt.show()

# 4. heatmap: which two columns measure the same thing?
sns.heatmap(dragons[["wingspan_m", "fire_temp_c", "age_years", "weight_t"]].corr(), annot=True)
plt.title("Correlation"); plt.show()
```

```python
print(dragons.groupby("species")[["wingspan_m", "fire_temp_c", "weight_t"]].mean().round(1))
print(dragons[["wingspan_m", "weight_t"]].corr().round(2))
```

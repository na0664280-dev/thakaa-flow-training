```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Sara", "Noura", "Reem", "Hind"],
    "city": ["Riyadh", "Jeddah", "Riyadh", "Dammam"],
    "score": [88, 92, 75, 60],
}, index=["a", "b", "c", "d"])

print(type(df["score"]))
print(type(df[["score"]]))
print(df.loc["b", "city"])
print(df.iloc[1, 1])
print(df.loc["a":"c", "score"].tolist())
print(df.iloc[0:2, 2].tolist())
print(df[df["score"] > 80]["name"].tolist())
print(df["city"].value_counts()["Riyadh"])
```

### الأسئلة

**س2: الفرق بين df["score"] و df[["score"]]**
`df["score"]` بقوسين واحد يرجع عمودًا واحدًا بنوع Series. أما `df[["score"]]` بقوسين مزدوجين فيرجع جدولًا من عمود واحد بنوع DataFrame. أحتاج الثاني لما أريد اختيار أكثر من عمود، أو أريد أن تبقى النتيجة جدولًا وليست عمودًا مفردًا.

**س3: لماذا loc["a":"c"] يرجع 3 صفوف و iloc[0:2] يرجع صفّين؟**
لأن `loc` يعتمد على أسماء الصفوف ويشمل آخر عنصر في النطاق (a و b و c). أما `iloc` فيعتمد على مواقع الصفوف ولا يشمل آخر رقم، فـ `0:2` يعطي الموقعين 0 و 1 فقط.

**س4: الفرق بين NumPy array و DataFrame، ولماذا نحتاج Pandas؟**
NumPy array مصفوفة أرقام بنوع بيانات واحد وبدون أسماء أعمدة. أما DataFrame فجدول فيه أسماء للأعمدة والصفوف، ويقبل أنواعًا مختلفة (نصوص وأرقام)، ويوفر دوال جاهزة للتنظيف والتجميع والتعامل مع القيم الناقصة. نحتاج Pandas لأن بيانات الواقع جداول مختلطة وفيها نقص، وNumPy وحدها لا تكفي لذلك.

**س5: جدول من حياتي (جدول الدراسة)**
- **الصف:** كل مادة أو كل يوم دراسة.
- **العمود:** اسم المادة، عدد ساعات المذاكرة، الدرجة، الملاحظات.
- **العمود الأكثر نقصًا:** الملاحظات، لأني لا أكتبها كل مرة وغالبًا أنساها.

```python
df = pd.DataFrame(data)

# 1) القيم الناقصة والمكررة (قبل التنظيف)
print(df.isna().sum())
print("Duplicated rows:", df.duplicated().sum())

# 2) حذف المكرر
df = df.drop_duplicates()
print("After drop:", df.shape)

# 3) توحيد المدن
df["city"] = df["city"].str.strip().str.title()
print(df["city"].value_counts(dropna=False))

# 4) تحويل السعر لرقم
df["price"] = df["price"].replace("eighteen", "18")
df["price"] = df["price"].astype(float)

# 5) التقييم المستحيل يصير ناقص
df.loc[(df["rating"] < 1) | (df["rating"] > 5), "rating"] = None

# 6) ملء الناقص
print("Quantity median:", df["quantity"].median())
df["quantity"] = df["quantity"].fillna(df["quantity"].median())
df["rating"] = df["rating"].fillna(df["rating"].median())
df["city"] = df["city"].fillna(df["city"].mode()[0])
df["drink"] = df["drink"].fillna("Unknown")

# 7) التحقق النهائي
print(df.isna().sum())
print(df)
```

```python
# 1) عمود الإيراد (كتب التمرين revenue، انتبه للإملاء)
df["revenue"] = df["price"] * df["quantity"]
print("Total revenue:", df["revenue"].sum())

# 2) إيراد كل مدينة، مرتب تنازليًا
print(df.groupby("city")["revenue"].sum().sort_values(ascending=False))

# 3) أكثر مشروب مبيعًا بالكمية + متوسط التقييم لكل مشروب
print(df.groupby("drink")["quantity"].sum().sort_values(ascending=False))
print(df.groupby("drink")["rating"].mean())

# 4) نسبة الطلبات من كل مدينة
print(df["city"].value_counts(normalize=True))

# 5) أعلى طلب من حيث الإيراد
print(df.loc[df["revenue"].idxmax()])
```

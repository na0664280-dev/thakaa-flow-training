**1. لماذا نرسم؟**
المجموعات الأربع لها نفس المتوسط والانحراف والارتباط تقريبًا، لكن رسوماتها مختلفة تمامًا. يعني الأرقام الملخِّصة وحدها قد تخفي القصة الحقيقية، والرسم يكشفها. مثال: معدلي 85 قد يعني أن كل موادي 85، أو أن نصفها 100 ونصفها 70. الرقم واحد والقصة مختلفة.

**2. اختر الرسمة**

| السؤال | الرسمة | لماذا؟ |
|---|---|---|
| كيف تغيرت مبيعات المقهى خلال 12 شهرًا؟ | Line | لأنها تتغير عبر الزمن |
| أي فرع باع أكثر من 5 فروع؟ | Bar | لمقارنة فئات منفصلة |
| كيف تتوزع أعمار 500 عميل؟ | Histogram | لتوزيع متغير رقمي واحد |
| هل يزيد الإيجار كلما زادت مساحة الشقة؟ | Scatter | لعلاقة بين متغيرين رقميين |
| هل في رواتب الموظفين قيم شاذة؟ | Histogram أو Box plot | لأنه يُظهر القيم البعيدة عن الباقي |

**3. أجزاء الرسمة**
- **Figure**: اللوحة الكاملة التي تحتوي كل شيء.
- **Axes**: منطقة الرسم نفسها (المحوران والبيانات) داخل الـ Figure.
- **العنوان**: يقول ماذا تعرض الرسمة.
- **عناوين المحاور**: توضح ما يقيسه كل محور ووحدته.
- **Legend**: يوضح أي لون أو خط يمثل أي شيء.
- لو أرسلتها لمديري بلا هذه الأجزاء، لن يفهم ماذا يرى ولن يستطيع الاعتماد عليها.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(2026)
sol = np.arange(1, 61)
temp = np.round(-60 + 15 * np.sin(sol / 9) + rng.normal(0, 4, 60), 1)
storm = np.zeros(60, dtype=int)
storm[[11, 12, 13, 32, 33, 46, 47, 48, 49]] = 1  # عاصفة ترابية = 1
temp[storm == 1] -= 12
wind = np.round(20 + storm * 35 + rng.normal(0, 5, 60), 1)
```

```python
plt.figure(figsize=(12, 5))
plt.plot(sol, temp, label="Temperature")
plt.scatter(sol[storm == 1], temp[storm == 1], color="red", label="Dust storm", zorder=3)
plt.axhline(temp.mean(), color="gray", linestyle="--", label="Mean")

coldest = temp.argmin()
plt.annotate(f"Coldest: Sol {sol[coldest]}, {temp[coldest]}°C",
             xy=(sol[coldest], temp[coldest]),
             xytext=(sol[coldest] + 3, temp[coldest] + 5),
             arrowprops=dict(arrowstyle="->"))

plt.title("Mars Temperature per Sol")
plt.xlabel("Sol")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
```

```python
fig, axs = plt.subplots(2, 1, figsize=(12, 7), sharex=True)

axs[0].plot(sol, temp)
axs[0].set_title("Temperature")
axs[0].set_ylabel("Temperature (°C)")

colors = ["red" if s == 1 else "steelblue" for s in storm]
axs[1].bar(sol, wind, color=colors)
axs[1].set_title("Wind Speed (red = storm)")
axs[1].set_xlabel("Sol")
axs[1].set_ylabel("Wind (km/h)")

fig.suptitle("Mars Station: Temperature and Wind")
plt.tight_layout()
plt.show()
```

```python
best_start, best_avg = None, -999
for i in range(len(sol) - 4):
    if storm[i:i+5].sum() == 0:
        avg = temp[i:i+5].mean()
        if avg > best_avg:
            best_avg, best_start = avg, i

print("Best window: Sol", sol[best_start], "to", sol[best_start + 4])
print("Average temp:", round(best_avg, 1))
```
